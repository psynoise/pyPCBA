# Optimize preferred component values

Use `e_series()` to define allowed resistor values, then use `optimize()` to rank resistor pairs for a voltage divider. The equations target a divider ratio of 0.33 and an equivalent parallel resistance of 50 Ohm.

```python
from sympy import Eq, symbols

from pyPCBA.lists import e_series
from pyPCBA.values import optimize
```

```python
# Voltage divider
resistors1 = e_series("E12", "1k", "100k")

R1, R2 = symbols("R1 R2")
eq1 = Eq(R2 / (R1 + R2), 0.33)

solutions = optimize(eq1, resistors1, num_solutions=5)
solutions
```

```python
# Voltage divider with 50 Ohm output resistance
resistors2 = e_series("E12", "1", "100k")

R1, R2 = symbols("R1 R2")
eqSys1 = [
    Eq(R2 / (R1 + R2), 0.33),
    Eq(R1 * R2 / (R1 + R2), 50),
]

solutions2 = optimize(eqSys1, resistors2, num_solutions=10)
solutions2
```
