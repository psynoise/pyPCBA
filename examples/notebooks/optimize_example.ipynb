# Optimize preferred component values

Use `e_series()` to define allowed resistor values, then use `optimize()` to rank resistor pairs for a voltage divider. The equations target a divider ratio of 0.33 and an equivalent parallel resistance of 50 Ohm.

```python
from sympy import Eq, symbols

from pyPCBA.lists import e_series
from pyPCBA.values import optimize
```

## Example 1 - Voltage divider with ratio 0.33

```python
resistors1 = e_series("E12", "1k", "100k")

R1, R2 = symbols("R1 R2")
eq1 = Eq(R2 / (R1 + R2), 0.33)

solutions1 = optimize(eq1, resistors1, num_solutions=5)
solutions1
```

## Example 2 - Voltage divider with 50 Ohm output resistance

```python
resistors2 = e_series("E12", "1", "100k")

R1, R2 = symbols("R1 R2")
eqSys2 = [
    Eq(R2 / (R1 + R2), 0.33),
    Eq(R1 * R2 / (R1 + R2), 50),
]

solutions2 = optimize(eqSys2, resistors2, num_solutions=10)
solutions2
```

## Example 3 - Using special operator "|" for voltage or current division

```python
# Special operators are translated by optimize() before SymPy parses them.
resistors3 = e_series("E12", "1k", "100k")

eq3 = "R1 | R2 = 0.33"

solutions3 = optimize(eq3, resistors3, num_solutions=5)
solutions3
```

## Example 4 - Using special operator "||" for output resistance calculation

```python
resistors4 = e_series("E12", "1", "100k")

R1, R2 = symbols("R1 R2")
eqSys4 = [
    "R1 | R2 = 0.33",
    "R1 || R2 = 50"
]

solutions4 = optimize(eqSys4, resistors4, num_solutions=10)
solutions4
```
